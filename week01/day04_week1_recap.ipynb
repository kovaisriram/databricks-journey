```python
%sql
create or replace table learn.sandbox.nation_region
select n.n_name as nation, r.r_name as region
from samples.tpch.nation n join samples.tpch.region r on n.n_regionkey = r.r_regionkey;

select region, count(distinct nation) as nations
from learn.sandbox.nation_region
group by region
order by region ;
```
